```python
# ================== Block 1 ==================
# 显示图像 + 交互点击记录点
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
import cv2
import json

# ================== 1. 读取图像 ==================
image_path = "/home/ubuntu/data/mmw/data_online/4/rgb/frame_20260305_003747_215388.png"

image = cv2.imread(image_path)
if image is None:
    raise FileNotFoundError(f"图像未找到: {image_path}")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("[INFO] 已加载图像，准备点击标注")

# ================== 2. 记录点 ==================
clicked_points = []
point_labels = []

# ================== 3. 画图函数 ==================
def show_points(coords, labels, ax, marker_size=200):
    pos = coords[labels==1]
    neg = coords[labels==0]

    if len(pos) > 0:
        ax.scatter(pos[:,0], pos[:,1], color='green', s=marker_size, marker='*')
    if len(neg) > 0:
        ax.scatter(neg[:,0], neg[:,1], color='red', s=marker_size, marker='*')

# ================== 4. 绘制图像 ==================
fig, ax = plt.subplots(figsize=(8,8))
ax.imshow(image)
ax.set_title("左键=正点(绿), 右键=负点(红).  点击完成后运行 Block 2")
ax.axis("off")

# ================== 5. 点击事件 ==================
def onclick(event):
    if event.xdata is None or event.ydata is None:
        return

    x, y = int(event.xdata), int(event.ydata)

    if event.button == 1:   # 左键
        clicked_points.append([x,y])
        point_labels.append(1)
        print(f"[CLICK] 正点: ({x}, {y})")
        ax.scatter(x, y, color='green', marker='*', s=200)
    elif event.button == 3: # 右键
        clicked_points.append([x,y])
        point_labels.append(0)
        print(f"[CLICK] 负点: ({x}, {y})")
        ax.scatter(x, y, color='red', marker='*', s=200)

    fig.canvas.draw()

# 绑定事件
cid = fig.canvas.mpl_connect("button_press_event", onclick)

print("[INFO] 你现在可以在图片上点击多个点。点击结束后生成 mask")
```

```python
# ================== Block 2 ==================
# 使用 Block 1 的 clicked_points / point_labels 进行 SAM 推理并保存最佳 mask

import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
import sys
import torch

# ================== 1. 检查点是否存在 ==================
if len(clicked_points) == 0:
    raise ValueError("你还没有在 Block 1 中点击任何点！")

# ================== 2. SAM 初始化 ==================
sys.path.append("..")
from segment_anything import sam_model_registry, SamPredictor

file_path = os.getcwd()                     # /data/mmw/test/segment-anything/notebooks
project_root = os.path.dirname(file_path)   # /data/mmw/test/segment-anything
project_root = os.path.dirname(project_root) # /data/mmw/test

sam_checkpoint = os.path.join(
    project_root,
    "segment-anything/checkpoints/sam_vit_h_4b8939.pth"
)

print("[INFO] 自动生成的 SAM 路径为：", sam_checkpoint)
# sam_checkpoint = "/data/mmw/test/segment-anything/checkpoints/sam_vit_h_4b8939.pth"
model_type = "vit_h"

device = "cuda"
sam = sam_model_registry[model_type](checkpoint=sam_checkpoint)
sam.to(device=device)

predictor = SamPredictor(sam)
predictor.set_image(image)

# 转 numpy
input_points = np.array(clicked_points)
input_labels = np.array(point_labels)

print("[INFO] 开始 SAM 多点推理 ...")

# ================== 3. SAM 推理 ==================
masks, scores, logits = predictor.predict(
    point_coords=input_points,
    point_labels=input_labels,
    multimask_output=True
)

# ================== 4. 选取最佳 mask ==================
best_idx = np.argmax(scores)
best_mask = masks[best_idx]
best_score = scores[best_idx]

print(f"[INFO] 最佳 mask index={best_idx}, score={best_score:.4f}")

# 保存路径
#save_path1 = "../../data_online/2/preprocess_image/best_mask.png"
save_path2 = "/home/ubuntu/data/mmw/data_online/4/masks/frame_20260305_003747_215388.png"
# 0/1 转 0/255
mask_to_save = (best_mask.astype(np.uint8) * 255)
#cv2.imwrite(save_path1, mask_to_save)
cv2.imwrite(save_path2, mask_to_save)
# print(f"[INFO] 最佳 Mask 已保存至: {save_path1}")
print(f"[INFO] 最佳 Mask 已保存至: {save_path2}")
print("[DEBUG] mask real path:", os.path.abspath(save_path2))


# ================== 5. 可视化最佳 mask ==================
# def show_mask(mask, ax, random_color=False):
#     color = np.array([30/255, 144/255, 255/255, 0.6])
#     h, w = mask.shape[-2:]
#     mask_image = mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
#     ax.imshow(mask_image)

# plt.figure(figsize=(10,10))
# plt.imshow(image)
# show_mask(best_mask, plt.gca())
# show_points(input_points, input_labels, plt.gca())
# plt.title(f"Best Mask (score={best_score:.3f})")
# plt.axis("off")
# plt.show()
# ================== 5. 使用 OpenCV 生成 overlay ==================

# overlay_save_path = "/home/ubuntu/data/mmw/data_online/1/masks/frame_20251225_010154_850624_overlay.png"

# # 如果 image 是 RGB（matplotlib常见），先转 BGR
# img_bgr = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

# # overlay 副本
# overlay = img_bgr.copy()

# # mask 转 bool
# mask = best_mask.astype(bool)

# # 覆盖颜色 (BGR)
# color = np.array([255, 144, 30], dtype=np.uint8)  # 蓝色

# # 透明度
# alpha = 0.6

# # alpha blend
# overlay[mask] = (
#     overlay[mask] * (1 - alpha) +
#     color * alpha
# ).astype(np.uint8)

# # ================== 6. 在图上画点击点 ==================

# for (x, y), label in zip(input_points, input_labels):
#     x, y = int(x), int(y)

#     if label == 1:   # 正样本点
#         cv2.circle(overlay, (x, y), 8, (0,255,0), -1)
#     else:            # 负样本点
#         cv2.circle(overlay, (x, y), 8, (0,0,255), -1)

# # ================== 7. 保存 ==================

# cv2.imwrite(overlay_save_path, overlay)

# print(f"[INFO] overlay mask 已保存至: {overlay_save_path}")
# print("[DEBUG] overlay real path:", os.path.abspath(overlay_save_path))
# ================== 5. 生成 overlay ==================
# ================== 5. 保存：半透明mask + 星形点击点 ==================

overlay_save_path = "/home/ubuntu/data/mmw/data_online/4/masks/frame_20260305_003747_215388_overlay_star.png"

# 1) 先用 OpenCV 做半透明 mask overlay（BGR空间）
img_bgr = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
overlay_bgr = img_bgr.copy()

mask = best_mask.astype(bool)

color_bgr = np.array([255, 144, 30], dtype=np.uint8)  # 蓝色(BGR)
alpha = 0.6
overlay_bgr[mask] = (overlay_bgr[mask] * (1 - alpha) + color_bgr * alpha).astype(np.uint8)

# 转回 RGB 供 matplotlib 显示/保存
overlay_rgb = cv2.cvtColor(overlay_bgr, cv2.COLOR_BGR2RGB)

# 2) 用 matplotlib 画“同款”五角星点并保存（marker='*'）
fig, ax = plt.subplots(figsize=(8, 8), dpi=200)
ax.imshow(overlay_rgb)
ax.axis("off")

# 复用 Block1 的 show_points（marker='*'）
input_points = np.array(clicked_points)
input_labels = np.array(point_labels)

def show_points(coords, labels, ax, marker_size=200):
    coords = np.asarray(coords)
    labels = np.asarray(labels)
    pos = coords[labels == 1]
    neg = coords[labels == 0]
    if len(pos) > 0:
        ax.scatter(pos[:, 0], pos[:, 1], color='green', s=marker_size, marker='*')
    if len(neg) > 0:
        ax.scatter(neg[:, 0], neg[:, 1], color='red', s=marker_size, marker='*')

show_points(input_points, input_labels, ax, marker_size=200)

# 保存最终可视化（无白边）
fig.savefig(overlay_save_path, bbox_inches="tight", pad_inches=0)
plt.close(fig)

print(f"[INFO] Overlay+MatplotlibStar 已保存至: {overlay_save_path}")
print("[DEBUG] real path:", os.path.abspath(overlay_save_path))
```
